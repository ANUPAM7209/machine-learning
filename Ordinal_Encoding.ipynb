```python
import numpy as np
import pandas as pd
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder , OrdinalEncoder
```

```python
df = pd.read_csv('covid_toy.csv')
```

```python
df.head(5)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['has_covid']),df['has_covid'],
                                                test_size=0.2 )
```

```python
X_train.head(5)
```

#Simple way

```python
si = SimpleImputer(strategy='mean')
X_train_si = si.fit_transform(X_train[['fever']])

X_test_si = si.transform(X_test[['fever']])

X_train_si.shape
```

```python
ohe = OrdinalEncoder(categories=[['Mild', 'Strong']])
X_train_cough = ohe.fit_transform(X_train[['cough']])

X_test_cough =  ohe.transform(X_train[['cough']])

X_train_cough.shape
```

```python
oe = OneHotEncoder(drop = 'first' , sparse_output = False)
X_train_gender_city = ohe.fit_transform(X_train[['gender','city']])

# also the test data
X_test_gender_city = ohe.transform(X_test[['gender','city']])

X_train_gender_city.shape
```
